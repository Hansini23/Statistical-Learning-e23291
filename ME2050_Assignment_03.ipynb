#Q1) Prediction of fraudulent transactions

A bank monitors credit-card transactions for possible fraud.

For each transaction, let the sample space be

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where:

* $\omega_1$: low-risk customer, legitimate transaction,
* $\omega_2$: low-risk customer, fraudulent transaction,
* $\omega_3$: high-risk customer, legitimate transaction,
* $\omega_4$: high-risk customer, fraudulent transaction.

Assume the probabilities are

$$
P(\{\omega_1\})=0.50,\qquad
P(\{\omega_2\})=0.10,\qquad
P(\{\omega_3\})=0.20,\qquad
P(\{\omega_4\})=0.20.
$$

Let $X:\Omega\to\mathbb R$ denote the fraud indicator:

$$
X(\omega_1)=0,\qquad
X(\omega_2)=1,\qquad
X(\omega_3)=0,\qquad
X(\omega_4)=1.
$$

Thus $X=1$ means the transaction is fraudulent, and $X=0$ means it is legitimate.

Now suppose the bank does not observe the full outcome $\omega$, but only the customer's risk class. Define $Y:\Omega\to\mathbb R$ by

$$
Y(\omega_1)=0,\qquad
Y(\omega_2)=0,\qquad
Y(\omega_3)=1,\qquad
Y(\omega_4)=1.
$$

Thus:

* $Y=0$ means “low-risk customer,”
* $Y=1$ means “high-risk customer.”

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

3. Explain why $E[X\mid Y]$ must be $\sigma(Y)$-measurable, and therefore must be constant on the sets $\{ \omega_1,\omega_2\}$ and $\{\omega_3,\omega_4\}$.

4. Compute $E[X\mid Y]$ explicitly and show that

$$
E[X\mid Y]= \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}
$$

5. Verify the defining property of conditional expectation by checking that for every set $A\in\sigma(Y)$,

$$
\int_A E[X\mid Y]\,dP=\int_A X\,dP.
$$

6. Find the Conditional PMF $p_{X|Y}(x \mid y)$.

7. Verify the identity:
$$\mathbb{E}[X\mid Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x \mid y)$$.

8. Calculate the **Mean Square Error (MSE)**

9. How would the **Mean Square Error (MSE)** change if the bank decided to ignore the customer risk class $Y$ and instead used the global average fraud rate ($E[X] = 0.30$) as a constant estimate for every transaction? Calculate this error and explain its relationship to the **Law of Total Variance**.

10. Using the joint distribution of $(X, Y)$ derived from the sample space:

    a) Compute the entropy $H(X)$, representing the bank's initial uncertainty regarding fraud.

    b) Compute the conditional entropy $H(X \mid Y)$, representing the uncertainty remaining after the customer's risk class is revealed.

    c) Calculate the Mutual Information $I(X; Y)$.

    d) Based on these values, what percentage of the total uncertainty about fraud is removed by knowing the customer's risk class?

1).

**Sample space:**
$$\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$$

**$\sigma$-algebra:** Since $\mathcal{F} = \mathcal{P}(\Omega)$ (the power set of $\Omega$), it contains all $2^4 = 16$ subsets:
$$\mathcal{F} = \{\emptyset,\ \{\omega_1\},\ \{\omega_2\},\ \{\omega_3\},\ \{\omega_4\},\ \{\omega_1,\omega_2\},\ \{\omega_1,\omega_3\},\ \{\omega_1,\omega_4\},\ \{\omega_2,\omega_3\},\ \{\omega_2,\omega_4\},\ \{\omega_3,\omega_4\},\ \{\omega_1,\omega_2,\omega_3\},\ \{\omega_1,\omega_2,\omega_4\},\ \{\omega_1,\omega_3,\omega_4\},\ \{\omega_2,\omega_3,\omega_4\},\ \Omega\}$$

**Probability measure:**
$$P(\{\omega_1\}) = 0.50, \quad P(\{\omega_2\}) = 0.10, \quad P(\{\omega_3\}) = 0.20, \quad P(\{\omega_4\}) = 0.20$$

For any event $A \in \mathcal{F}$, the probability is computed by summing over the atoms:
$$P(A) = \sum_{\omega_i \in A} P(\{\omega_i\})$$

**Verification:**
$$0.50 + 0.10 + 0.20 + 0.20 = 1.00 \checkmark$$

2). The $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$, is found by identifying the level sets (preimages) of $Y$:

$$Y^{-1}(\{0\}) = \{\omega \in \Omega : Y(\omega) = 0\} = \{\omega_1, \omega_2\} \quad \text{(low-risk customers)}$$

$$Y^{-1}(\{1\}) = \{\omega \in \Omega : Y(\omega) = 1\} = \{\omega_3, \omega_4\} \quad \text{(high-risk customers)}$$

$\sigma(Y)$ is the smallest $\sigma$-algebra that makes $Y$ measurable. It must contain these level sets and be closed under complements and countable unions:

$$\sigma(Y) = \{\emptyset,\ \{\omega_1,\omega_2\},\ \{\omega_3,\omega_4\},\ \Omega\}$$

This is a coarse $\sigma$-algebra containing only $4$ sets, capturing exactly the information: **"which risk class does the customer belong to?"**

3).
$\mathbb{E}[X \mid Y]$ must be $\sigma(Y)$-measurable because by the definition of conditional expectation, it is the unique function measurable with respect to $\sigma(Y)$ satisfying:

$$\int_A \mathbb{E}[X \mid Y] \, dP = \int_A X \, dP \quad \forall A \in \sigma(Y)$$

The atoms of $\sigma(Y)$ are:

$$A_0 = \{\omega_1, \omega_2\} \quad \text{(low-risk, } Y=0\text{)}$$
$$A_1 = \{\omega_3, \omega_4\} \quad \text{(high-risk, } Y=1\text{)}$$

Since $\mathbb{E}[X \mid Y]$ can only use the information encoded in $Y$, it cannot distinguish between outcomes within the same atom. Therefore it must be constant on each atom:

$$\mathbb{E}[X \mid Y](\omega_1) = \mathbb{E}[X \mid Y](\omega_2) = c_0$$

$$\mathbb{E}[X \mid Y](\omega_3) = \mathbb{E}[X \mid Y](\omega_4) = c_1$$

4).
 Find $P(Y=0)$ and $P(Y=1)$:

$$P(Y=0) = P(\{\omega_1, \omega_2\}) = 0.50 + 0.10 = 0.60$$

$$P(Y=1) = P(\{\omega_3, \omega_4\}) = 0.20 + 0.20 = 0.40$$

Compute the constant on each atom:

$$c_0 = \mathbb{E}[X \mid Y=0] = \frac{\int_{\{\omega_1,\omega_2\}} X \, dP}{P(Y=0)} = \frac{X(\omega_1)\cdot P(\{\omega_1\}) + X(\omega_2)\cdot P(\{\omega_2\})}{0.60}$$

$$= \frac{0 \cdot 0.50 + 1 \cdot 0.10}{0.60} = \frac{0.10}{0.60} = \frac{1}{6}$$

$$c_1 = \mathbb{E}[X \mid Y=1] = \frac{\int_{\{\omega_3,\omega_4\}} X \, dP}{P(Y=1)} = \frac{X(\omega_3)\cdot P(\{\omega_3\}) + X(\omega_4)\cdot P(\{\omega_4\})}{0.40}$$

$$= \frac{0 \cdot 0.20 + 1 \cdot 0.20}{0.40} = \frac{0.20}{0.40} = \frac{1}{2}$$

**Therefore:**

$$\mathbb{E}[X \mid Y] = \frac{1}{6}\mathbf{1}_{\{Y=0\}} + \frac{1}{2}\mathbf{1}_{\{Y=1\}}$$

5).
We verify that $\int_A \mathbb{E}[X \mid Y] \, dP = \int_A X \, dP$ for every $A \in \sigma(Y)$:

**Case 1:** $A = \emptyset$

$$\int_{\emptyset} \mathbb{E}[X \mid Y] \, dP = 0 = \int_{\emptyset} X \, dP \quad \checkmark$$

**Case 2:** $A = \{\omega_1, \omega_2\}$

$$\int_{\{\omega_1,\omega_2\}} \mathbb{E}[X \mid Y] \, dP = \frac{1}{6} \cdot 0.50 + \frac{1}{6} \cdot 0.10 = \frac{1}{6} \cdot 0.60 = 0.10$$

$$\int_{\{\omega_1,\omega_2\}} X \, dP = 0 \cdot 0.50 + 1 \cdot 0.10 = 0.10 \quad \checkmark$$

**Case 3:** $A = \{\omega_3, \omega_4\}$

$$\int_{\{\omega_3,\omega_4\}} \mathbb{E}[X \mid Y] \, dP = \frac{1}{2} \cdot 0.20 + \frac{1}{2} \cdot 0.20 = \frac{1}{2} \cdot 0.40 = 0.20$$

$$\int_{\{\omega_3,\omega_4\}} X \, dP = 0 \cdot 0.20 + 1 \cdot 0.20 = 0.20 \quad \checkmark$$

**Case 4:** $A = \Omega$

$$\int_{\Omega} \mathbb{E}[X \mid Y] \, dP = 0.10 + 0.20 = 0.30$$

$$\int_{\Omega} X \, dP = 0 \cdot 0.50 + 1 \cdot 0.10 + 0 \cdot 0.20 + 1 \cdot 0.20 = 0.30 \quad \checkmark$$

The defining property holds for all $A \in \sigma(Y)$.

6).
Using Bayes' rule: $p_{X|Y}(x \mid y) = \dfrac{P(X=x,\ Y=y)}{P(Y=y)}$

**Given $Y=0$ (low-risk customers):**

$$p_{X|Y}(0 \mid 0) = \frac{P(\{\omega_1\})}{P(Y=0)} = \frac{0.50}{0.60} = \frac{5}{6}$$

$$p_{X|Y}(1 \mid 0) = \frac{P(\{\omega_2\})}{P(Y=0)} = \frac{0.10}{0.60} = \frac{1}{6}$$

**Given $Y=1$ (high-risk customers):**

$$p_{X|Y}(0 \mid 1) = \frac{P(\{\omega_3\})}{P(Y=1)} = \frac{0.20}{0.40} = \frac{1}{2}$$

$$p_{X|Y}(1 \mid 1) = \frac{P(\{\omega_4\})}{P(Y=1)} = \frac{0.20}{0.40} = \frac{1}{2}$$

**Verification:**

$$\sum_{x \in \{0,1\}} p_{X|Y}(x \mid 0) = \frac{5}{6} + \frac{1}{6} = 1 \quad \checkmark$$

$$\sum_{x \in \{0,1\}} p_{X|Y}(x \mid 1) = \frac{1}{2} + \frac{1}{2} = 1 \quad \checkmark$$

---

7).
**Given $y=0$:**

$$\mathbb{E}[X \mid Y=0] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x \mid 0)$$

$$= 0 \cdot \frac{5}{6} + 1 \cdot \frac{1}{6} = \frac{1}{6} \quad \checkmark$$

**Given $y=1$:**

$$\mathbb{E}[X \mid Y=1] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x \mid 1)$$

$$= 0 \cdot \frac{1}{2} + 1 \cdot \frac{1}{2} = \frac{1}{2} \quad \checkmark$$

Both values match the conditional expectation computed in Part 4:

$$\mathbb{E}[X \mid Y] = \frac{1}{6}\mathbf{1}_{\{Y=0\}} + \frac{1}{2}\mathbf{1}_{\{Y=1\}} \quad \checkmark$$

---

8).
The MSE of the conditional expectation $\mathbb{E}[X \mid Y]$ is:

$$\text{MSE} = \mathbb{E}\left[(X - \mathbb{E}[X \mid Y])^2\right] = \sum_{i=1}^{4} \left(X(\omega_i) - \mathbb{E}[X \mid Y](\omega_i)\right)^2 \cdot P(\{\omega_i\})$$

**At each outcome:**

$$\omega_1: \left(0 - \frac{1}{6}\right)^2 \cdot 0.50 = \frac{1}{36} \cdot 0.50 = \frac{0.50}{36}$$

$$\omega_2: \left(1 - \frac{1}{6}\right)^2 \cdot 0.10 = \frac{25}{36} \cdot 0.10 = \frac{2.50}{36}$$

$$\omega_3: \left(0 - \frac{1}{2}\right)^2 \cdot 0.20 = \frac{1}{4} \cdot 0.20 = \frac{1.80}{36}$$

$$\omega_4: \left(1 - \frac{1}{2}\right)^2 \cdot 0.20 = \frac{1}{4} \cdot 0.20 = \frac{1.80}{36}$$

**Summing all contributions:**

$$\text{MSE} = \frac{0.50 + 2.50 + 1.80 + 1.80}{36} = \frac{6.60}{36} = \frac{11}{60} \approx 0.1833$$

---

9).
If the bank ignores $Y$ and uses the global average $\mathbb{E}[X] = 0.30$ as a constant estimate:

$$\text{MSE}_{\text{flat}} = \mathbb{E}\left[(X - 0.30)^2\right] = \text{Var}(X)$$

**Computing $\text{Var}(X)$:**

$$\mathbb{E}[X^2] = 0^2 \cdot 0.70 + 1^2 \cdot 0.30 = 0.30$$

$$\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = 0.30 - (0.30)^2 = 0.30 - 0.09 = 0.21$$

**Law of Total Variance:**

$$\text{Var}(X) = \underbrace{\mathbb{E}[\text{Var}(X \mid Y)]}_{\text{MSE of } \mathbb{E}[X \mid Y]} + \underbrace{\text{Var}(\mathbb{E}[X \mid Y])}_{\text{variance explained by } Y}$$

**Computing $\text{Var}(\mathbb{E}[X \mid Y])$:**

$$\mathbb{E}[\mathbb{E}[X \mid Y]] = \frac{1}{6} \cdot 0.60 + \frac{1}{2} \cdot 0.40 = 0.10 + 0.20 = 0.30$$

$$\mathbb{E}\left[(\mathbb{E}[X \mid Y])^2\right] = \left(\frac{1}{6}\right)^2 \cdot 0.60 + \left(\frac{1}{2}\right)^2 \cdot 0.40 = \frac{0.60}{36} + \frac{0.40}{4} = \frac{1}{60} + \frac{1}{10} = \frac{7}{60}$$

$$\text{Var}(\mathbb{E}[X \mid Y]) = \frac{7}{60} - (0.30)^2 = \frac{7}{60} - \frac{9}{100} = \frac{70}{600} - \frac{54}{600} = \frac{16}{600} = \frac{2}{75} \approx 0.0267$$

**Verification:**

$$\mathbb{E}[\text{Var}(X \mid Y)] + \text{Var}(\mathbb{E}[X \mid Y]) = \frac{11}{60} + \frac{2}{75} = \frac{55}{300} + \frac{8}{300} = \frac{63}{300} = 0.21 = \text{Var}(X) \quad \checkmark$$

**Comparison:**

$$\text{MSE}_{\text{flat}} = 0.21 \quad \text{(ignoring } Y\text{)}$$

$$\text{MSE}_{\text{cond}} = \frac{11}{60} \approx 0.1833 \quad \text{(using } Y\text{)}$$

$$\text{Reduction} = 0.21 - 0.1833 = 0.0267 \approx 12.7\%$$

Knowing the customer's risk class $Y$ reduces the MSE by $\text{Var}(\mathbb{E}[X \mid Y]) = \dfrac{2}{75} \approx 0.0267$, which is exactly the variance explained by $Y$.

---

10).
**a) Entropy $H(X)$:**

$$P(X=0) = P(\{\omega_1, \omega_3\}) = 0.50 + 0.20 = 0.70$$

$$P(X=1) = P(\{\omega_2, \omega_4\}) = 0.10 + 0.20 = 0.30$$

$$H(X) = -P(X=0)\log_2 P(X=0) - P(X=1)\log_2 P(X=1)$$

$$= -0.70\log_2(0.70) - 0.30\log_2(0.30)$$

$$= -0.70 \times (-0.5146) - 0.30 \times (-1.7370)$$

$$= 0.3602 + 0.5211 = \mathbf{0.8813} \text{ bits}$$

**b) Conditional Entropy $H(X \mid Y)$:**

$$H(X \mid Y=0) = -\frac{5}{6}\log_2\frac{5}{6} - \frac{1}{6}\log_2\frac{1}{6}$$

$$= -\frac{5}{6}(-0.2630) - \frac{1}{6}(-2.5850) = 0.2192 + 0.4308 = 0.6500 \text{ bits}$$

$$H(X \mid Y=1) = -\frac{1}{2}\log_2\frac{1}{2} - \frac{1}{2}\log_2\frac{1}{2} = 1.0000 \text{ bit}$$

$$H(X \mid Y) = P(Y=0)\cdot H(X \mid Y=0) + P(Y=1)\cdot H(X \mid Y=1)$$

$$= 0.60 \times 0.6500 + 0.40 \times 1.0000 = 0.3900 + 0.4000 = \mathbf{0.7900} \text{ bits}$$

**c) Mutual Information $I(X;Y)$:**

$$I(X;Y) = H(X) - H(X \mid Y) = 0.8813 - 0.7900 = \mathbf{0.0913} \text{ bits}$$

**d) Percentage of uncertainty removed:**

$$\frac{I(X;Y)}{H(X)} \times 100 = \frac{0.0913}{0.8813} \times 100 \approx \mathbf{10.36\%}$$

Knowing the customer's risk class $Y$ removes approximately $\mathbf{10.36\%}$ of the total uncertainty about fraud. While $Y$ is statistically informative, it explains only a modest fraction of the total entropy, reflecting that fraud occurs across both risk categories.

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

#Answer



## Step 1: Partition the Joint Density

The joint density is

$$
f_{X,Y}(x,y)
=
\frac{1}
{(2\pi)^{(n+m)/2}|\Sigma|^{1/2}}
\exp\left(
-\frac{1}{2}
(z-\mu)^T
\Sigma^{-1}
(z-\mu)
\right)
$$

where

$$
z=
\begin{bmatrix}
x \\
y
\end{bmatrix},
\qquad
\mu=
\begin{bmatrix}
\mu_X \\
\mu_Y
\end{bmatrix}
$$

and

$$
\Sigma=
\begin{bmatrix}
\Sigma_{XX} & \Sigma_{XY} \\
\Sigma_{YX} & \Sigma_{YY}
\end{bmatrix}
$$


## Step 2: Block Matrix Inversion of \(\Sigma^{-1}\)

Using the Schur complement, the inverse of the partitioned covariance matrix is

$$
\Sigma^{-1}
=
\begin{bmatrix}
\Sigma_{X|Y}^{-1}
&
-\Sigma_{X|Y}^{-1}\Sigma_{XY}\Sigma_{YY}^{-1}

$$

$$
\Sigma_{X|Y}

\Sigma_{XX}
-
\Sigma_{XY}
\Sigma_{YY}^{-1}
\Sigma_{YX}
$$



## Step 3: Expand the Quadratic Form

Let

$$
\tilde{x}=x-\mu_X
\qquad \text{and} \qquad
\tilde{y}=y-\mu_Y
$$

Then the exponent expands as

$$

\Sigma^{-1}
(z-\mu)

\Sigma_{X|Y}^{-1}
\tilde{x}

\Sigma_{X|Y}^{-1}
\Sigma_{XY}
\Sigma_{YY}^{-1}
\tilde{y}
$$

$$
+
\tilde{y}^T
\left(
\Sigma_{YY}^{-1}
+
\Sigma_{YY}^{-1}
\Sigma_{YX}
\Sigma_{X|Y}^{-1}
\Sigma_{XY}
\Sigma_{YY}^{-1}
\right)
\tilde{y}
$$



## Step 4: Complete the Square in \(x\)

Define

$$
\mu^*
=
\mu_X
+
\Sigma_{XY}
\Sigma_{YY}^{-1}
(y-\mu_Y)
$$

Then

$$
(z-\mu)^T
\Sigma^{-1}

\Sigma_{X|Y}^{-1}

$$

$$



## Step 5: Substitute into the Joint Density

$$
f_{X,Y}(x,y)

## Step 7: Extract the Conditional Density

The marginal density of \(Y\) is

f_Y(y)

$$
f_{X|Y}(x \mid y)
=

Hence,

$$
\boxed{

# Q3) Continuous Fault Monitoring of a Spindle Motor

A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**
1.  **Hidden Temperature ($X$):** The standardized temperature deviation follows a standard normal distribution:
    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

---

**Answer the followings:**

**1. Joint Distribution**
Identify the distribution of the random varaible $(X, Y)$.
* Show that $Y \sim \mathscr{N}(0, 5)$.
* Find the covariance $\text{Cov}(X, Y)$.

**2. Information Geometry**
Explain why $X$ is **not** $\sigma(Y)$-measurable. In practical terms, if the power meter shows $Y = 5$, why can't the engineer know for certain if the motor is overheating ($X$ is high) or if the material being cut is simply very hard ($Q$ is high)?

2).

$X$ is **not** $\sigma(Y)$-measurable because knowing $Y$ does not uniquely determine $X$.

Since $Y = X + Q$, the same observed value of $Y$ can come from infinitely many
combinations of $X$ and $Q$.

**Example — if the power meter shows $Y = 5$:**

| Scenario | $X$ (Temperature) | $Q$ (Mechanical Load) | Conclusion |
|----------|-------------------|----------------------|------------|
| 1 | $4$ | $1$ | Motor overheating, light load |
| 2 | $1$ | $4$ | Motor normal, heavy load |
| 3 | $0$ | $5$ | Motor fine, very heavy cut |

All three give $Y = 5$, but represent completely different physical states.

 The $\sigma$-algebra generated by $Y$,

$$\sigma(Y) = \{\{Y \in B\} : B \in \mathcal{B}(\mathbb{R})\}$$

contains only events expressible purely as $\{Y \in B\}$. Since $X$ depends on
the unobservable $Q$, it cannot be expressed as a function of $Y$ alone.

Observing $Y$ gives *partial* information about $X$,
but the mechanical noise $Q$ creates irreducible uncertainty.

7).

The factory manager says:

> “Since \(Y=X+Q\), our best estimate of temperature should just be \(Y\) itself.”

However, this is not correct because \(Y\) contains both the true temperature \(X\) and the mechanical noise \(Q\).

From Part 3, the conditional expectation is

$$
E[X|Y=y]
=
\frac{\operatorname{Cov}(X,Y)}
{\operatorname{Var}(Y)}
y
$$

Since

$$
\operatorname{Cov}(X,Y)=1
\qquad \text{and} \qquad
\operatorname{Var}(Y)=5
$$

we obtain

$$
E[X|Y=y]
=
\frac{1}{5}y
$$

Therefore,

$$
\boxed{
E[X|Y]=\frac{Y}{5}
}
$$

This shows that the best estimate is not \(Y\), but only one-fifth of \(Y\).

The reason is that most of the variation in \(Y\) comes from the mechanical noise \(Q\), whose variance is 4, while the temperature variance is only 1.

Thus, the observation \(Y\) exaggerates the actual temperature changes.

The estimator \(E[X|Y]\) “shrinks” the measurement toward the mean value \(0\) because it removes the effect of noise and keeps only the part of \(Y\) that is statistically related to \(X\).

Hence, \(Y\) alone is a noisy and biased estimator of the true temperature, while \(E[X|Y]\) is the optimal minimum mean square error (MMSE) estimate.
