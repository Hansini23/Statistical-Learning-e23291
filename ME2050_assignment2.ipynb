#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



1). Probability Space (Ω, F, P)

$$\Omega = \{0, 1, 2, 3\}$$

$$\mathcal{F} = 2^\Omega = \{\emptyset, \{0\}, \{1\}, \{2\}, \{3\}, \{0,1\}, \{0,2\}, \ldots, \{0,1,2,3\}\} \quad (16 \text{ subsets})$$

$$P(\{0\})=0.50, \quad P(\{1\})=0.30, \quad P(\{2\})=0.15, \quad P(\{3\})=0.05$$

$$P(A) = \sum_{\omega \in A} P(\{\omega\}), \qquad P(\Omega) = 0.50+0.30+0.15+0.05 = 1 $$

2). X as a Random Variable

$$X : \Omega \to \mathbb{R}, \qquad X(\omega) = \omega$$

$$X(0)=0, \quad X(1)=1, \quad X(2)=2, \quad X(3)=3$$

X is measurable since every subset of the finite space $\Omega$ belongs to $\mathcal{F}$.

3). Probability Mass Function

$$p_X(x) = P(X = x) = \begin{cases} 0.50, & x = 0, \\ 0.30, & x = 1, \\ 0.15, & x = 2, \\ 0.05, & x = 3, \\ 0, & \text{otherwise.} \end{cases}$$



 4). E[X] = Σ x · P(X = x)

$E[X] = \sum x \cdot P(X = x)$

$= 0 \cdot (0.50) + 1 \cdot (0.30) + 2 \cdot (0.15) + 3 \cdot (0.05)$

$= 0 + 0.30 + 0.30 + 0.15$

$= {0.75}$

5). Var(X) = E[X²] − (E[X])²
        = 1.35 − (0.75)²
        = 1.35 − 0.5625
        = 0.7875



6). P(X ≥ 1) = 1 − P(X = 0)
           = 1 − 0.50
           = 0.50

   P(X ≥ 2) = P(X = 2) + P(X = 3)
           = 0.15 + 0.05
           = 0.20


 7). Practical Meaning of Expectation and Variance

Expectation E[X] = 0.75

On average, there are 0.75 transformer failures per day — about 3 failures every 4 days. This helps the district plan staff and spare parts.

Variance Var(X) = 0.7875

The variance is close to the mean, meaning the number of failures varies quite a lot from day to day. Some days have zero failures, while other days may have 2 or 3. This means the district must always keep extra resources ready for high-failure days.

# Q4) Probability of Defective Items

In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

2. Define $X$ as a continuous random variable on this space.

3. Write down the probability density function of $X$.

4. Verify that $f$ is a valid probability density function.

5. Compute the expectation $E[X]$.

6. Compute the variance $\mathrm{Var}(X)$.

7. Find the probability that the defective proportion is at most $0.10$, that is, compute

$$
P(X\le 0.10).
$$

You may leave your answer in Beta integral form.

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute

$$
P(0.05\le X\le 0.15).
$$

You may leave your answer in integral form.

9. Briefly explain why a Beta distribution is a reasonable model for this situation.

10. Interpret the expectation and variance in the context of production quality.


 1). Probability Space (Ω, F, P)

$$\Omega = [0, 1] \qquad \text{(all possible defective proportions)}$$

$$\mathcal{F} = \mathcal{B}([0,1]) \qquad \text{(Borel σ-algebra on } [0,1]\text{)}$$

$$P(A) = \int_A f(x)\, dx \qquad \text{for all } A \in \mathcal{F}$$


2). X as a Continuous Random Variable

$$X : \Omega \to [0,1], \qquad X(\omega) = \omega$$

X represents the **proportion of defective items** produced on a given day.


3). Probability Density Function

$$f(x) = \begin{cases} \dfrac{1}{B(2,18)}\, x^{1}(1-x)^{17}, & 0 < x < 1, \\[6pt] 0, & \text{otherwise.} \end{cases}$$

where the Beta function is:

$$B(2,18) = \frac{\Gamma(2)\,\Gamma(18)}{\Gamma(20)} = \frac{1! \cdot 17!}{19!} = \frac{1}{190}$$

So the PDF simplifies to:

$$f(x) = 190\, x(1-x)^{17}, \qquad 0 < x < 1$$

 4). Verification that f is a Valid PDF

Condition 1: Non-negativity

$$f(x) = 190\, x(1-x)^{17} \geq 0 \qquad \text{for all } x \in [0,1] $$

Condition 2: Integrates to 1

$$\int_0^1 f(x)\,dx = 190 \int_0^1 x^{2-1}(1-x)^{18-1}\,dx = 190 \cdot B(2,18)$$

$$= 190 \times \frac{1}{190} = 1 $$


5). Expectation E[X]

For a Beta(α, β) distribution:

$$E[X] = \frac{\alpha}{\alpha + \beta} = \frac{2}{2 + 18} = \frac{2}{20} = \boxed{0.10}$$

On average, 10% of items produced are defective.

6). Variance Var(X)

For a Beta(α, β) distribution:

$$\text{Var}(X) = \frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)} = \frac{2 \times 18}{(20)^2(21)} = \frac{36}{8400} = \boxed{0.00\overline{428}}$$

$$\text{Var}(X) \approx 0.0043$$

7). P(X ≤ 0.10)

$$P(X \leq 0.10) = \int_0^{0.10} 190\, x(1-x)^{17}\, dx$$

In Beta integral form:

$$= \frac{B_{0.10}(2,\,18)}{B(2,\,18)} = I_{0.10}(2,\,18)$$

where $I_x(\alpha,\beta)$ is the regularised incomplete Beta function.

$$P(X \leq 0.10) \approx 0.5691$$

8). P(0.05 ≤ X ≤ 0.15)

$$P(0.05 \leq X \leq 0.15) = \int_{0.05}^{0.15} 190\, x(1-x)^{17}\, dx$$

$$= I_{0.15}(2,18) - I_{0.05}(2,18)$$

$$\approx 0.7981 - 0.2265 = \boxed{0.5716}$$

9). Why Beta Distribution is a Reasonable Model
- X is a proportion, so it naturally lives in $[0,1]$ — the Beta distribution is defined exactly on this interval.
- The Beta distribution is flexible: its shape changes with α and β, allowing skewed or symmetric fits.
- With α=2, β=18, the distribution is right-skewed, placing most probability near 0 — realistic for a good manufacturing process where defects are rare.
- It is the conjugate prior for binomial proportions, making it statistically natural for modelling defect rates.


10). Interpretation of E[X] and Var(X) in Production Quality

E[X] = 0.10

The process produces 10% defective items on average. This is the long-run defect rate the quality team should plan for — useful for setting acceptance thresholds and estimating waste.

Var(X) ≈ 0.0043,  SD(X) ≈ 0.066

The defect proportion is relatively stable from day to day (SD ≈ 6.6 percentage points). Most days the defect rate stays close to 10%, which means the process is consistent, though management should still monitor for days when the rate climbs toward 15–20%.

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

1. Specify the support of $X$.

2. Write down the probability density function $f_X(x)$.

3. Verify that $f_X$ integrates to $1$ over its support.

4. Compute the expectation $E[X]$.

5. Compute the variance $\mathrm{Var}(X)$.

6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

7. Explain briefly why a Triangular distribution is a reasonable model in this setting.

8. Compute the probability that the installation takes at most $5$ hours, that is, compute

$$
P(X \leq 5).
$$

9. Compute the probability that the installation takes between $5$ and $8$ hours, that is, compute

$$
P(5 \leq X \leq 8).
$$

10. Briefly explain how this model could be used in project planning or sensitivity analysis.


1). Support of X

$$\text{Support of } X = [a,\, b] = [4,\, 10]$$

X takes values between 4 hours(minimum) and 10 hours (maximum).


2). Probability Density Function

For a Triangular distribution with minimum $a=4$, maximum $b=10$, mode $c=6$:

$$f_X(x) = \begin{cases} \dfrac{2(x-a)}{(b-a)(c-a)}, & a \leq x < c \\[10pt] \dfrac{2}{b-a}, & x=c\\[10pt]\dfrac{2(b-x)}{(b-a)(b-c)}, & c < x \leq b \\[10pt] 0, & \text{otherwise} \end{cases}$$

Substituting $a=4,\ b=10,\ c=6$:

$$f_X(x) = \begin{cases} \dfrac{2(x-4)}{(6)(2)} = \dfrac{x-4}{6}, & 4 \leq x \leq 6 \\[10pt] \dfrac{2}{10-4}=\dfrac{1}{3}, &x=6\\[10pt]\dfrac{2(10-x)}{(6)(4)} = \dfrac{10-x}{12}, & 6 < x \leq 10 \\[10pt] 0, & \text{otherwise} \end{cases}$$

3). Verification that $f_X$ Integrates to 1

$$\int_4^{10} f_X(x)\,dx = \int_4^6 \frac{x-4}{6}\,dx + \int_6^{10} \frac{10-x}{12}\,dx$$

**First integral:**

$$\int_4^6 \frac{x-4}{6}\,dx = \frac{1}{6} \cdot \frac{(x-4)^2}{2}\Bigg|_4^6 = \frac{1}{6} \cdot \frac{4}{2} = \frac{1}{3}$$

**Second integral:**

$$\int_6^{10} \frac{10-x}{12}\,dx = \frac{1}{12} \cdot \left[10x - \frac{x^2}{2}\right]_6^{10} = \frac{1}{12} \cdot 8 = \frac{2}{3}$$

**Total:**

$$\frac{1}{3} + \frac{2}{3} = 1$$

 4). Expectation E[X]

For a Triangular distribution:

$$E[X] = \frac{a + b + c}{3} = \frac{4 + 10 + 6}{3} = \frac{20}{3} \approx {6.67 \text{ hours}}$$

5). Variance Var(X)

For a Triangular distribution:

$$\text{Var}(X) = \frac{a^2 + b^2 + c^2 - ab - ac - bc}{18}$$

$$= \frac{16 + 100 + 36 - 40 - 24 - 60}{18} = \frac{28}{18} = {\frac{14}{9} \approx 1.556 \text{ hours}^2}$$



6). Interpretation of a, b, and c

| Parameter | Value | Meaning |
|-----------|-------|---------|
| $a = 4$ | 4 hours | Minimum — the fastest the installation could possibly be completed under ideal conditions |
| $b = 10$ | 10 hours | Maximum — the longest time if everything goes wrong |
| $c = 6$ | 6 hours | Mode — the most likely completion time based on expert judgment |


7). Why Triangular Distribution is Reasonable

- When no historical data is available, experts can easily estimate minimum, maximum, and most likely times — exactly the three parameters needed.
- It is bounded, so it never predicts impossibly short or long durations.
- It is intuitive and transparent — easy to explain to project managers and stakeholders.
- It naturally allows asymmetry: here $c=6$ is closer to $a=4$ than to $b=10$, reflecting that delays are more likely than early finishes.


8). P(X ≤ 5)

Since $4 \leq 5 \leq 6$, we use the first piece:

$$P(X \leq 5) = \int_4^5 \frac{x-4}{6}\,dx = \frac{1}{6} \cdot \frac{(x-4)^2}{2}\Bigg|_4^5 = \frac{1}{6} \cdot \frac{1}{2} = {\frac{1}{12} \approx 0.0833}$$

There is only an 8.33% chance the installation finishes within 5 hours.

9). P(5 ≤ X ≤ 8)

Split at the mode $c = 6$:

$$P(5 \leq X \leq 8) = \int_5^6 \frac{x-4}{6}\,dx + \int_6^8 \frac{10-x}{12}\,dx$$

**First part:**

$$\int_5^6 \frac{x-4}{6}\,dx = \frac{1}{6} \cdot \frac{(x-4)^2}{2}\Bigg|_5^6 = \frac{1}{12}(4 - 1) = \frac{3}{12} = \frac{1}{4}$$

**Second part:**

$$\int_6^8 \frac{10-x}{12}\,dx = \frac{1}{12}\left[10x - \frac{x^2}{2}\right]_6^8 = \frac{1}{12}(48 - 42) = \frac{6}{12} = \frac{1}{2}$$

**Total:**

$$P(5 \leq X \leq 8) = \frac{1}{4} + \frac{1}{2} = {\frac{3}{4} = 0.75}$$

There is a 75% chance the installation takes between 5 and 8 hours.

10). Use in Project Planning and Sensitivity Analysis

Project Planning:
- The expected completion time $E[X] \approx 6.67$ hours can be used to schedule labour, equipment, and hospital downtime around the installation.
- $P(X \leq 5) \approx 8.3\%$ tells planners that finishing early is unlikely — they should not plan other tasks to start before 7–8 hours.

Sensitivity Analysis:
- Planners can adjust $a$, $b$, or $c$ to model optimistic, pessimistic, or realistic scenarios and see how the expected time and probabilities change.
- For example, if supply delays push $a$ to 5 hours, a new distribution gives a revised $E[X]$ and risk profile instantly.
- This makes the Triangular model ideal for PERT (Program Evaluation and Review Technique) in construction and engineering projects.

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$

8. Compute the approximate standard deviation $\sigma$.

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.


1). Shape Parameters α and β

$$\alpha = 1 + 4\left(\frac{c - a}{b - a}\right) = 1 + 4\left(\frac{11 - 8}{20 - 8}\right) = 1 + 4\left(\frac{3}{12}\right) = 1 + 1 = {2}$$

$$\beta = 1 + 4\left(\frac{b - c}{b - a}\right) = 1 + 4\left(\frac{20 - 11}{20 - 8}\right) = 1 + 4\left(\frac{9}{12}\right) = 1 + 3 = {4}$$


2). Support of X

$$\text{Support of } X = [a,\, b] = [8,\, 20]$$

X takes values between 8 days (minimum) and 20 days (maximum).

3). Probability Density Function

Substituting $\alpha = 2,\ \beta = 4,\ a = 8,\ b = 20$:

$$f_X(x) = \frac{(x - 8)^{1}(20 - x)^{3}}{B(2,\,4)\cdot(12)^{5}}, \qquad 8 \leq x \leq 20$$

where:

$$B(2, 4) = \frac{\Gamma(2)\,\Gamma(4)}{\Gamma(6)} = \frac{1! \cdot 3!}{5!} = \frac{6}{120} = \frac{1}{20}$$

Therefore:

$${f_X(x) = \frac{20(x-8)(20-x)^3}{12^5} = \frac{20(x-8)(20-x)^3}{248832}}, \qquad 8 \leq x \leq 20$$

and

$${f_X(x) = 0} \text{   otherwise}$$

4). Probability Space (Ω, F, P)

$$\Omega = [8,\, 20] \qquad \text{(all possible completion times in days)}$$

$$\mathcal{F} = \mathcal{B}([8, 20]) \qquad \text{(Borel } \sigma\text{-algebra on } [8, 20]\text{)}$$

$$P(A) = \int_A f_X(x)\, dx \qquad \text{for all } A \in \mathcal{F}$$


5). X as a Random Variable

$$X : \Omega \to [8,\, 20], \qquad X(\omega) = \omega$$

X is measurable since the pre-image of every Borel set in $\mathbb{R}$ belongs to $\mathcal{F}$.  
X represents the number of days to complete the software integration task.

6). Derivation of E[X]

The PERT distribution is a scaled Beta on $[a, b]$. Using the substitution $t = \dfrac{x - a}{b - a}$, so that $x = a + (b-a)t$:

$$E[X] = a + (b - a)\cdot E[T], \qquad T \sim \text{Beta}(\alpha, \beta)$$

$$E[T] = \frac{\alpha}{\alpha + \beta}$$

The PERT parameters give:

$$\alpha + \beta = \left(1 + 4\cdot\frac{c-a}{b-a}\right) + \left(1 + 4\cdot\frac{b-c}{b-a}\right) = 2 + \frac{4(b-a)}{b-a} = 6$$

$$\alpha = 1 + \frac{4(c-a)}{b-a} \implies (b-a)\,\alpha = (b-a) + 4(c-a)$$

Therefore:

$$E[X] = a + (b-a)\cdot\frac{\alpha}{6} = a + \frac{(b-a) + 4(c-a)}{6} = a + \frac{b - a + 4c - 4a}{6}$$

$${E[X] = \frac{a + 4c + b}{6}}$$

Substituting values:

$$E[X] = \frac{8 + 4(11) + 20}{6} = \frac{8 + 44 + 20}{6} = \frac{72}{6} = {12 \text{ days}}$$

7). Variance Var(X)

Using the PERT approximation:

$$\text{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7} = \frac{(12 - 8)(20 - 12)}{7} = \frac{4 \times 8}{7} = \frac{32}{7} \approx {4.571 \text{ days}^2}$$

8). Standard Deviation σ

$$\sigma = \sqrt{\text{Var}(X)} = \sqrt{\frac{32}{7}} \approx {2.138 \text{ days}}$$


 9). PERT vs Triangular Distribution

| Feature | Triangular | PERT |
|---|---|---|
| $E[X]$ formula | $\dfrac{a + c + b}{3}$ | $\dfrac{a + 4c + b}{6}$ |
| Weight on mode $c$ | Equal weight $\frac{1}{3}$ | Heavier weight $\frac{4}{6}$ |
| Shape | Piecewise linear (sharp peak) | Smooth Beta curve |
| Sensitivity to mode | Less sensitive | More sensitive |

**For this problem:**

$$E[X]_{\text{Triangular}} = \frac{8 + 11 + 20}{3} = 13 \text{ days}$$

$$E[X]_{\text{PERT}} = \frac{8 + 4(11) + 20}{6} = 12 \text{ days}$$

The PERT distribution gives more weight to the expert's mode estimate $c = 11$, pulling the mean closer to it. The Triangular distribution treats all three parameters equally, making it more sensitive to extreme values $a$ and $b$. In practice, the mode is usually the most reliable expert estimate, so PERT is more appropriate.

10). Practical Interpretation

| Value | Result | Meaning |
|---|---|---|
| $a = 8$ days | Minimum | Best-case scenario — everything goes perfectly |
| $b = 20$ days | Maximum | Worst-case scenario — major delays or issues |
| $c = 11$ days | Mode | Most likely duration based on expert judgment |
| $E[X] = 12$ days | Expected | The project manager should plan for **12 days** on average |
| $\sigma \approx 2.14$ days | Std. deviation | On most days the task finishes within **12 ± 2.14 days** |

**In project planning:**
- Schedule the task with a baseline of 12 days
- Set a buffer of ≈ 2–3 days(one standard deviation) to absorb uncertainty
- The probability of finishing within 14 days (one σ above mean) covers the majority of scenarios
- The long right tail toward $b = 20$ warns planners to have contingency plans for worst-case delays

#Q7) If the PERT Expectation is to Match Our Intuition

A project manager models the duration $X$ of a task by a PERT distribution with parameters $a$, $b$, and $c$, where $a$ is the minimum time, $b$ is the maximum time, and $c$ is the most likely time.

Suppose the minimum time is $a=5$ days and the most likely time is $c=11$ days. If the expected time computed from the PERT model is also to be the same as the most likely time:

1. Determine the value of the maximum time $b$.

2. What happens to the variance?







Given:
$$a = 5, \qquad c = 11, \qquad E[X] = c = 11$$

The PERT expectation formula is:

$$E[X] = \frac{a + 4c + b}{6}$$



 1).

Finding b

Set $E[X] = c = 11$:

$$\frac{a + 4c + b}{6} = 11$$

$$a + 4c + b = 66$$

Substituting $a = 5,\ c = 11$:

$$5 + 4(11) + b = 66$$

$$5 + 44 + b = 66$$

$$49 + b = 66$$

$${b = 17 \text{ days}}$$


 2).

Using the PERT variance approximation:

$$\text{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7}$$

Substituting $E[X] = 11,\ a = 5,\ b = 17$:

$$\text{Var}(X) = \frac{(11 - 5)(17 - 11)}{7} = \frac{6 \times 6}{7} = \frac{36}{7} \approx{5.143 \text{ days}^2}$$

$$\sigma = \sqrt{\frac{36}{7}} \approx {2.27 \text{ days}}$$

**Observation:**

When $E[X] = c$, the expected value sits exactly at the mode. This forces:

$$E[X] - a = b - E[X] = 6$$

meaning the mean is equidistant from both $a$ and $b$ — the distribution becomes symmetric around the mode. The variance is non-zero because uncertainty still exists between the bounds $[5, 17]$, but the distribution is now balanced with no skew.
